```python
import pandas as pd

transaction = pd.read_csv(r"C:\Users\Adith\OneDrive\Desktop\fedbank\data\train_transaction.csv")
identity = pd.read_csv(r"C:\Users\Adith\OneDrive\Desktop\fedbank\data\train_identity.csv")

df = transaction.merge(identity, on="TransactionID", how="left")
```

```python
v_report = pd.DataFrame({
    "Feature": v_features,
    "Null_Count": [df[col].isna().sum() for col in v_features],
    "Null_Percentage": [
        df[col].isna().mean() * 100
        for col in v_features
    ],
    "Unique_Values": [
        df[col].nunique(dropna=True)
        for col in v_features
    ],
    "Data_Type": [
        str(df[col].dtype)
        for col in v_features
    ]
})

v_report = v_report.sort_values(
    "Null_Percentage",
    ascending=False
)

print(v_report.to_string(index=False))
```

```python
# V-features
v_features = [f"V{i}" for i in range(1, 340)]
v_features = [c for c in v_features if c in df.columns]

# Select V-features with >5% missing values
v_indicator_features = [
    c for c in v_features
    if df[c].isna().mean() > 0.05
]

# Create all indicators at once
missing_indicators = pd.DataFrame(
    {
        f"{col}_missing": df[col].isna().astype("int8")
        for col in v_indicator_features
    },
    index=df.index
)

# Add them all at once
df = pd.concat([df, missing_indicators], axis=1)

# Defragment
df = df.copy()

print("V-features:", len(v_features))
print("Missing indicators created:", len(v_indicator_features))
print("New shape:", df.shape)
```

```python
import numpy as np
import pandas as pd

# Check infinite values
inf_count = np.isinf(df[v_features].to_numpy()).sum()

# Check extremely large values
very_large_count = (
    df[v_features].abs() > 1e30
).sum().sum()

print("Inf values:", inf_count)
print("Very large values (>1e30):", very_large_count)

# Examine largest values
v_max = (
    df[v_features]
    .max()
    .sort_values(ascending=False)
)

print("\nTop 20 maximum values:")
print(v_max.head(20))
```

```python
# Calculate missing percentage for every V-feature
v_missing_rates = df[v_features].isna().mean() * 100

# 50–90% missing → median
v_median_features = v_missing_rates[
    (v_missing_rates >= 50) & (v_missing_rates <= 90)
].index.tolist()

# <50% missing → KNN
v_knn_features = v_missing_rates[
    v_missing_rates < 50
].index.tolist()

print("Median imputation features:", len(v_median_features))
print("KNN imputation features:", len(v_knn_features))

print("\nMedian features:")
print(v_median_features)

print("\nKNN features:")
print(v_knn_features)
```

```python
# ============================================
# STEP 1: MEDIAN IMPUTATION FOR 50–90% MISSING
# ============================================

for col in v_median_features:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)

print("Median imputation completed.")

# Verify
remaining_median_missing = df[v_median_features].isna().sum().sum()

print("Remaining missing values in median features:",
      remaining_median_missing)
```

```python
# ============================================
# STEP 2: CHECK KNN FEATURES
# ============================================

knn_report = pd.DataFrame({
    "Feature": v_knn_features,
    "Missing_Count": [df[c].isna().sum() for c in v_knn_features],
    "Missing_Percentage": [df[c].isna().mean() * 100 for c in v_knn_features],
    "Unique_Values": [df[c].nunique(dropna=True) for c in v_knn_features],
    "Dtype": [str(df[c].dtype) for c in v_knn_features]
})

knn_report = knn_report.sort_values(
    "Missing_Percentage",
    ascending=False
)

print(knn_report.to_string(index=False))
```

```python
print("\nTotal missing values in KNN features:",
      df[v_knn_features].isna().sum().sum())

print("\nRows containing at least one missing KNN feature:",
      df[v_knn_features].isna().any(axis=1).sum())

print("\nRows with no missing KNN features:",
      df[v_knn_features].notna().all(axis=1).sum())
```

```python
# ============================================
# PREPARE KNN REFERENCE DATA
# ============================================

import numpy as np
import pandas as pd

# Rows that have no missing values among the 180 KNN features
complete_mask = df[v_knn_features].notna().all(axis=1)

complete_indices = df.index[complete_mask]

print("Complete reference rows available:", len(complete_indices))

# Use a reproducible sample of 20,000 complete rows
RANDOM_STATE = 42
REFERENCE_SIZE = min(20000, len(complete_indices))

reference_indices = (
    pd.Series(complete_indices)
    .sample(
        n=REFERENCE_SIZE,
        random_state=RANDOM_STATE
    )
    .values
)

knn_reference = df.loc[
    reference_indices,
    v_knn_features
].copy()

print("KNN reference shape:", knn_reference.shape)
print(
    "Missing values in reference:",
    knn_reference.isna().sum().sum()
)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
import numpy as np

# ============================================
# PREPARE STANDARDIZED KNN REFERENCE
# ============================================

# Convert reference data to float32 to reduce RAM
reference_array = knn_reference.to_numpy(dtype=np.float32)

print("Reference array shape:", reference_array.shape)
print(
    "Reference memory:",
    f"{reference_array.nbytes / (1024**2):.1f} MB"
)

# Standardize features
scaler = StandardScaler()

reference_scaled = scaler.fit_transform(
    reference_array
).astype(np.float32)

print("Standardized reference shape:", reference_scaled.shape)

# Build nearest-neighbor index
knn = NearestNeighbors(
    n_neighbors=5,
    metric="euclidean",
    algorithm="brute",
    n_jobs=-1
)

knn.fit(reference_scaled)

print("KNN reference index created successfully.")
```

```python
# ============================================================
# FEDBANK - V-FEATURE KNN IMPUTATION
# ============================================================
#
# 180 V-features with <50% missingness
# KNN imputation: k=5
# Reference set: 20,000 complete rows
# Batch processing to reduce RAM usage
#
# ============================================================

import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
import gc
import time
import os

# ============================================================
# SETTINGS
# ============================================================

K = 5
REFERENCE_SIZE = 20_000
BATCH_SIZE = 2_000
RANDOM_STATE = 42

print("=" * 70)
print("FEDBANK - KNN IMPUTATION")
print("=" * 70)

# ============================================================
# 1. CHECK REQUIRED VARIABLES
# ============================================================

print("\nChecking variables...")

required_variables = [
    "df",
    "v_features",
    "v_knn_features",
    "knn_reference"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Required variable '{variable}' was not found."
        )

print("All required variables found.")

print("\nDataset shape:", df.shape)
print("Total V-features:", len(v_features))
print("KNN V-features:", len(v_knn_features))
print("Reference rows:", len(knn_reference))

# ============================================================
# 2. VERIFY KNN FEATURES
# ============================================================

if len(v_knn_features) != 180:
    raise ValueError(
        f"Expected 180 KNN features, "
        f"but found {len(v_knn_features)}."
    )

print("\n✓ 180 KNN features confirmed.")

# ============================================================
# 3. CHECK CURRENT MISSING VALUES
# ============================================================

missing_before = (
    df[v_knn_features]
    .isna()
    .sum()
    .sum()
)

print(
    "\nMissing KNN values before imputation:",
    f"{missing_before:,}"
)

# ============================================================
# 4. FIND ROWS REQUIRING IMPUTATION
# ============================================================

print("\nFinding rows requiring KNN imputation...")

missing_mask = (
    df[v_knn_features]
    .isna()
    .any(axis=1)
)

rows_to_impute = df.index[
    missing_mask
].to_numpy()

print(
    "Rows requiring imputation:",
    f"{len(rows_to_impute):,}"
)

print(
    "Rows already complete:",
    f"{(~missing_mask).sum():,}"
)

# ============================================================
# 5. PREPARE COMPLETE REFERENCE DATA
# ============================================================

print("\nPreparing KNN reference data...")

# Make sure reference contains only KNN features
reference_array = (
    knn_reference[v_knn_features]
    .to_numpy(dtype=np.float32)
)

print(
    "Reference shape:",
    reference_array.shape
)

print(
    "Reference memory:",
    f"{reference_array.nbytes / (1024**2):.2f} MB"
)

# Verify reference has no missing values
reference_missing = np.isnan(
    reference_array
).sum()

print(
    "Missing values in reference:",
    reference_missing
)

if reference_missing != 0:
    raise ValueError(
        "Reference data contains missing values. "
        "KNN reference rows must be complete."
    )

# ============================================================
# 6. CREATE KNN IMPUTER
# ============================================================

print("\nCreating KNNImputer...")

imputer = KNNImputer(
    n_neighbors=K,
    weights="uniform",
    metric="nan_euclidean"
)

# ============================================================
# 7. FIT KNN IMPUTER ON 20,000 COMPLETE ROWS
# ============================================================

print(
    f"\nFitting KNNImputer on "
    f"{len(reference_array):,} reference rows..."
)

start_fit = time.time()

imputer.fit(reference_array)

fit_time = time.time() - start_fit

print(
    f"KNNImputer fitted successfully "
    f"in {fit_time / 60:.2f} minutes."
)

# ============================================================
# 8. PROCESS TARGET ROWS IN BATCHES
# ============================================================

total_rows = len(rows_to_impute)

total_batches = int(
    np.ceil(total_rows / BATCH_SIZE)
)

print("\n" + "=" * 70)
print("STARTING BATCH IMPUTATION")
print("=" * 70)

print("Total rows:", f"{total_rows:,}")
print("Batch size:", f"{BATCH_SIZE:,}")
print("Total batches:", f"{total_batches:,}")
print("K:", K)

start_total = time.time()

# Keep track of completed rows
completed_rows = 0

for batch_number, start in enumerate(
    range(0, total_rows, BATCH_SIZE),
    start=1
):

    end = min(
        start + BATCH_SIZE,
        total_rows
    )

    batch_indices = rows_to_impute[
        start:end
    ]

    batch_start_time = time.time()

    print(
        f"\n[{batch_number}/{total_batches}] "
        f"Rows {start:,} → {end:,}"
    )

    # --------------------------------------------------------
    # Extract batch
    # --------------------------------------------------------

    batch = (
        df.loc[
            batch_indices,
            v_knn_features
        ]
        .to_numpy(dtype=np.float32)
    )

    missing_in_batch_before = np.isnan(
        batch
    ).sum()

    print(
        "Missing values in batch:",
        f"{missing_in_batch_before:,}"
    )

    # --------------------------------------------------------
    # KNN IMPUTATION
    # --------------------------------------------------------

    batch_imputed = imputer.transform(
        batch
    )

    # --------------------------------------------------------
    # Convert result to float32
    # --------------------------------------------------------

    batch_imputed = batch_imputed.astype(
        np.float32,
        copy=False
    )

    # --------------------------------------------------------
    # Verify batch
    # --------------------------------------------------------

    missing_in_batch_after = np.isnan(
        batch_imputed
    ).sum()

    if missing_in_batch_after != 0:
        raise RuntimeError(
            f"Batch {batch_number} still contains "
            f"{missing_in_batch_after:,} missing values."
        )

    # --------------------------------------------------------
    # Put imputed values back into dataframe
    # --------------------------------------------------------

    df.loc[
        batch_indices,
        v_knn_features
    ] = batch_imputed

    # --------------------------------------------------------
    # Progress information
    # --------------------------------------------------------

    batch_time = (
        time.time()
        - batch_start_time
    )

    completed_rows = end

    elapsed = (
        time.time()
        - start_total
    )

    percentage = (
        completed_rows
        / total_rows
        * 100
    )

    print(
        f"✓ Batch completed in "
        f"{batch_time:.2f} seconds"
    )

    print(
        f"Progress: "
        f"{completed_rows:,}/{total_rows:,} "
        f"({percentage:.2f}%)"
    )

    print(
        f"Total elapsed: "
        f"{elapsed / 60:.2f} minutes"
    )

    # --------------------------------------------------------
    # Cleanup
    # --------------------------------------------------------

    del batch
    del batch_imputed

    gc.collect()

# ============================================================
# 9. FINAL VERIFICATION
# ============================================================

print("\n" + "=" * 70)
print("VERIFYING KNN IMPUTATION")
print("=" * 70)

remaining_knn_missing = (
    df[v_knn_features]
    .isna()
    .sum()
    .sum()
)

remaining_v_missing = (
    df[v_features]
    .isna()
    .sum()
    .sum()
)

print(
    "\nRemaining missing KNN values:",
    f"{remaining_knn_missing:,}"
)

print(
    "Remaining missing V values:",
    f"{remaining_v_missing:,}"
)

# ============================================================
# 10. FEATURE-BY-FEATURE VERIFICATION
# ============================================================

if remaining_knn_missing == 0:

    print(
        "\n✓ All 180 KNN features contain "
        "no missing values."
    )

else:

    print(
        "\n⚠ Some KNN features still contain "
        "missing values."
    )

    remaining_report = (
        df[v_knn_features]
        .isna()
        .sum()
    )

    remaining_report = (
        remaining_report[
            remaining_report > 0
        ]
        .sort_values(
            ascending=False
        )
    )

    print(remaining_report)

# ============================================================
# 11. CHECK MEDIAN FEATURES AGAIN
# ============================================================

remaining_median_missing = (
    df[v_median_features]
    .isna()
    .sum()
    .sum()
)

print(
    "\nRemaining missing median-imputed values:",
    f"{remaining_median_missing:,}"
)

# ============================================================
# 12. FINAL V-FEATURE CHECK
# ============================================================

print("\n" + "=" * 70)
print("FINAL V-FEATURE STATUS")
print("=" * 70)

print(
    "Total V-features:",
    len(v_features)
)

print(
    "KNN features:",
    len(v_knn_features)
)

print(
    "Median features:",
    len(v_median_features)
)

print(
    "Remaining V missing values:",
    f"{remaining_v_missing:,}"
)

# ============================================================
# 13. SAVE COMPLETED V-FEATURE DATA
# ============================================================

if remaining_v_missing == 0:

    output_file = (
        "fedbank_v_features_imputed.pkl"
    )

    print(
        f"\nSaving completed dataframe to:"
        f"\n{output_file}"
    )

    df.to_pickle(
        output_file
    )

    file_size_gb = (
        os.path.getsize(output_file)
        / (1024**3)
    )

    print(
        f"✓ Saved successfully."
    )

    print(
        f"File size: "
        f"{file_size_gb:.2f} GB"
    )

else:

    print(
        "\n⚠ Data was NOT saved as completed "
        "because missing V-values remain."
    )

# ============================================================
# 14. MEMORY CLEANUP
# ============================================================

del reference_array
del imputer
del missing_mask
del rows_to_impute

gc.collect()

# ============================================================
# 15. FINAL SUMMARY
# ============================================================

total_time = (
    time.time()
    - start_total
)

print("\n" + "=" * 70)
print("KNN IMPUTATION FINISHED")
print("=" * 70)

print(
    f"Total KNN processing time: "
    f"{total_time / 60:.2f} minutes"
)

print(
    "Final dataframe shape:",
    df.shape
)

print(
    "Remaining V missing values:",
    f"{df[v_features].isna().sum().sum():,}"
)

print("=" * 70)
```
