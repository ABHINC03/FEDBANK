```python
import pandas as pd

transaction = pd.read_csv(r"data\train_transaction.csv")
identity = pd.read_csv(r"data\train_identity.csv")

df = transaction.merge(identity, on="TransactionID", how="left")
```

```python
for col in df.columns:
    print(col)
print(len(df.columns))
```

```python

null_report = pd.DataFrame({
    "Column": df.columns,
    "Null_Count": df.isnull().sum(),
    "Null_Percentage": df.isnull().mean() * 100,
    "Data_Type": df.dtypes.astype(str),
    "Unique_Values": df.nunique(dropna=True)
})

null_report = null_report.sort_values(
    "Null_Percentage",
    ascending=False
)

print(null_report.to_string(index=False))
```

```python
v_features = [f"V{i}" for i in range(1, 340)]

# Keep only columns that actually exist
v_features = [col for col in v_features if col in df.columns]

print("Number of V-features:", len(v_features))
print(v_features)
```

```python
v_report = pd.DataFrame({
    "Feature": v_features,
    "Null_Count": [df[col].isna().sum() for col in v_features],
    "Null_Percentage": [
        df[col].isna().mean() * 100
        for col in v_features
    ],
    "Unique_Values": [
        df[col].nunique(dropna=True)
        for col in v_features
    ],
    "Data_Type": [
        str(df[col].dtype)
        for col in v_features
    ]
})

v_report = v_report.sort_values(
    "Null_Percentage",
    ascending=False
)

print(v_report.to_string(index=False))
```

```python
print("\n>90% missing:")
print((v_report["Null_Percentage"] > 90).sum())

print("\n50–90% missing:")
print(
    (
        (v_report["Null_Percentage"] >= 50) &
        (v_report["Null_Percentage"] <= 90)
    ).sum()
)

print("\n<50% missing:")
print((v_report["Null_Percentage"] < 50).sum())

print("\n>5% missing:")
print((v_report["Null_Percentage"] > 5).sum())
```
